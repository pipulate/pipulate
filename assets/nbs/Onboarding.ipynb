## 🎩 See Magic Trick 

**INSTRUCTIONS**: Press the `Esc` key, then type `0`, `0`. 

Normally resetting the Notebook's kernel would wipe its memory entirely, but Pipulate makes it remember. Run the next cell to see how the `wand` recovers your state. This mechanism is the foundation for turning these explicit, top-to-bottom Notebook workflows into robust web applications.

```python
from pipulate import wand
from imports import onboard_sauce as sauce

JOB_ID = wand.db['active_job']

# Recover the pointers from the SQLite disk!
recovered_name = wand.get(JOB_ID, "operator_name", "Alice")
recovered_url = wand.get(JOB_ID, "target_url", "https://example.com")

# Sync the pipeline state to the global FastHTML UI state
wand.db['operator_name'] = recovered_name

try:
    # Test if the Python variable survived
    _ = VOLATILE_CANARY 
    # If we get here, they DID NOT reset the kernel.
    wand.speak(
        f"Ah, you didn't reset the kernel! That's okay, but you missed the magic trick. \n"
        f"I still have your name: {recovered_name}. \n"
        f"And our target site is locked in[: {recovered_url}.] \n"
        "[Your data is safe ]on the disk."
    )
except NameError:
    # If we get here, the canary is dead. The kernel was successfully reset!
    wand.speak(
        f"Good work! You reset the kernel for the demonstration. \n"
        f"Even though the volatile memory was wiped, I recovered your name: {recovered_name}. \n"
        f"And our target site is locked in[: {recovered_url}]. \n"
        "While your data is safe on the disk, you can update it in the earlier step."
    )

wand.imperio()
```

## 🌐 Scrape Page

```python
from pipulate import wand
from imports import onboard_sauce as sauce

JOB_ID = wand.db['active_job']
recovered_url = wand.get(JOB_ID, "target_url", "https://example.com/")
wand.speak(
    f"Initializing browser automation for {recovered_url}. Hands off the mouse! \n"
    "Wait for the browser to close itself. This could take up to 30 seconds. \n"
    "Be patient — we are waiting out an invisible CAPTCHA to prove to the server \n"
    "that you are a carbon-based lifeform. \n\n",
    delay=1.5,     # <-- Give the browser a second to pop up before talking
    wait=False     # <-- Non-blocking! The cell moves immediately to wand.scrape
)

result = await wand.scrape(
    url=recovered_url, 
    headless=False,
    override_cache=True
)

if result.get('success'):
    if result.get('cached'):
        wand.speak("Cache Hit! Using existing artifacts. If you want to see the browser pop up again, change override_cache to True.")
    else:
        wand.speak("Scrape Successful.")
else:
    wand.speak("I encountered an error during navigation.")
    print(f"Scrape Failed: {result.get('error')}")

wand.imperio()
```
