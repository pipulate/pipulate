# 👀 Give Machine Eyes

## 🌐 Scrape Page

```python
from pipulate import wand
from imports import onboard_sauce as sauce

JOB_ID = wand.db['active_job']
recovered_url = wand.get(JOB_ID, "target_url", "https://example.com/")

if sauce.is_fast_track(JOB_ID):
    wand.speak("Fast-track active. Skipping the live scrape.")
    print(f"⏭️ Fast-track: scrape of {recovered_url} skipped.")
else:
    wand.speak("Initializing browser automation. Please wait...", wait=False)
    result = await wand.scrape(url=recovered_url, headless=False, override_cache=True)
    if result.get('success'):
        wand.speak("Cache Hit." if result.get('cached') else "Scrape Successful.")
    else:
        wand.speak("I encountered an error during navigation.")
        print(f"Scrape Failed: {result.get('error')}")

wand.imperio()
```

## 🎭 Persona

```python
from imports import onboard_sauce as sauce
from pipulate import wand

JOB_ID = wand.db.get('active_job', 'onboarding_01')
recovered_url = globals().get('recovered_url') or wand.get(JOB_ID, "target_url", "https://example.com/")
local_model = ACTIVE_MODEL if 'ACTIVE_MODEL' in globals() else None

if sauce.is_fast_track(JOB_ID):
    wand.speak("Fast-track active. Skipping the persona prompt draft.")
    print("⏭️ Fast-track: persona / prompt draft skipped.")
elif not local_model:
    wand.speak("No local AI available. Skipping the local prompt draft.")
    print("⏭️ No local model — hand-write your prompt below if you like.")
    sauce.render_prompt_workbench(JOB_ID, recovered_url)
else:
    wand.speak("Prompting local AI")
    draft_content = sauce.prepare_prompt_draft(JOB_ID, recovered_url, local_model)
    print(f"🤖 Local AI ({local_model}) has drafted your instructions.", end="\n\n")
    wand.speak("Local AI has drafted your instructions. Polish them below.")
    wand.copy_button(draft_content)
    sauce.render_prompt_workbench(JOB_ID, recovered_url)

wand.imperio(side_quest=True)
```
